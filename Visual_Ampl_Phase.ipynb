```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file_path = '/Users/ranmate/Downloads/amplitude_full.csv'

img = pd.read_csv(file_path, header=None).values

#нормирую легенду
img = img - img.min() 
img = img / img.max()

# обрезаю область – задаю порог
threshold = 0.05   # 5% от максимума

mask = img > threshold

# нахожу координаты всех ярких пикселей
ys, xs = np.where(mask)

# нахожу границы нужной области
x_min, x_max = xs.min(), xs.max()
y_min, y_max = ys.min(), ys.max()

# добавляю небольшой запас вокруг области
dop = 20

x_min = max(x_min - dop, 0)
x_max = min(x_max + dop, img.shape[1] - 1)

y_min = max(y_min - dop, 0)
y_max = min(y_max + dop, img.shape[0] - 1)

# обрезаю изображение
cropped = img[y_min:y_max + 1, x_min:x_max + 1]

plt.figure(figsize=(9, 7))

plt.imshow(
    cropped,
    cmap="viridis",
    vmin=0,
    vmax=1,
    extent=[x_min, x_max, y_max, y_min]
)

plt.title("Амплитуда", fontsize=24)
plt.xlabel("X пиксели", fontsize=20)
plt.ylabel("Y пиксели", fontsize=20)

plt.colorbar()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =========================
# 1. Чтение данных
# =========================
file_path = '/Users/ranmate/Downloads/phase_full.csv'   # путь к csv
phase = pd.read_csv(file_path, header=None, dtype=np.float32).values

# =========================
# 2. Если фаза в радианах: переводим в единицы pi
# =========================
phase_pi = phase / np.pi

# Ограничиваем диапазон [-1, 1]
phase_pi = np.clip(phase_pi, -1, 1)

# =========================
# 3. Построение
# =========================
fig, ax = plt.subplots(figsize=(10, 7))

im = ax.imshow(
    phase_pi,
    cmap="twilight_shifted",   # вот нужный colormap
    origin="lower",
    aspect="equal",
    vmin=-1,
    vmax=1,
    interpolation="nearest"
)

ax.set_title("Фазовая карта", fontsize=26)
ax.set_xlabel("X пиксели", fontsize=24)
ax.set_ylabel("Y пиксели", fontsize=24)

h, w = phase_pi.shape
ax.set_xticks([0, w//5, 2*w//5, 3*w//5, 4*w//5, w-1])
ax.set_yticks([0, h//5, 2*h//5, 3*h//5, 4*h//5, h-1])
ax.tick_params(axis='both', labelsize=14)

cbar = fig.colorbar(im, ax=ax)
cbar.set_ticks([-1, 1])
cbar.set_label("Фаза (π)", fontsize=22, rotation=90, labelpad=20)
cbar.ax.tick_params(labelsize=16)

plt.tight_layout()
plt.show()
```
